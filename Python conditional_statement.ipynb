```python
#10. Take a year as input and check whether it is a leap year.

year = 2027
if year % 400 == 0:
    print("Leap Year")
elif year % 100 == 0:
    print("Not a Leap Year")
elif year % 4 == 0:
    print("Leap Year")
else:
    print("Not a Leap Year")
```

```python
#11. Take marks and assign grade:  90 and above → A  75 to 89 → B  50 to 74 → C  Below 50 → Fail

marks = int(input("Enter your marks: "))
if marks >= 90:
    print("Grade A")
elif marks >= 75 and marks <= 89:
    print("Grade B")
elif marks >= 50 and marks <= 74:
    print("Grade C")
else:
    print("Fail")
```

```python
#12. Take a character as input and check whether it is a vowel or consonant.

ch = input("Enter a character: ").lower()
if ch == 'a' or ch == 'e' or ch == 'i' or ch == 'o' or ch == 'u':
    print("Vowel")
else:
    print("Consonant")
```

```python
#15. Create a simple calculator:
# Take two numbers and an operator (+, -, *, /)
# Perform operation based on user input.



a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
calculate = input("Enter operator (+, -, *, /): ")
if calculate == '+':
    print("Result =", a + b)
elif calculate == '-':
    print("Result =", a - b)
elif calculate == '*':
    print("Result =", a * b)
elif calculate == '/':
     print("Result =", a / b)
else:
    print("Invalid operator")
```
